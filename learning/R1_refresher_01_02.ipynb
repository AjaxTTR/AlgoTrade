## Part B — Reassignment and `+=` (the parked refresher)

Once a variable exists, you can put a new value in it. The old value is gone.

```python
balance = 100000
balance = balance + 500     # the right side uses the OLD balance, then writes the result back
```

Read `balance = balance + 500` as: *balance becomes (current balance plus 500)*. Python evaluates the right side first using the existing value, then assigns the result to the same name.

**Augmented assignment** is shorthand for that pattern:

| Long form | Short form | Read it as |
|---|---|---|
| `x = x + 5` | `x += 5` | x increases by 5 |
| `x = x - 5` | `x -= 5` | x decreases by 5 |
| `x = x * 2` | `x *= 2` | x doubles |
| `x = x / 2` | `x /= 2` | x halves |

All of these work the same way: evaluate the right side using the current value of `x`, write the result back to `x`. They are *not* magic — they're just typing-savers.

````python
# Predict the final value of `x` BEFORE running.
x = 10
x += 5      # x is now ?
x -= 2      # x is now ?
x *= 3      # x is now ?
x /= 2      # x is now ? — and what TYPE will it be?
print(x)
print(type(x))
````

Note the type at the end. `/` always returns a float, even if the result has no decimal part. This is why `x /= 2` switches `x` from int to float.

## Part C — Lists, indexing, slicing, `.append`

**List:** an ordered sequence of values, written with `[ ]` and commas: `pnls = [120, -80, 250, -50]`.

**Indexing** starts at `0`. Negative indices count from the end.

```python
pnls = [120, -80, 250, -50]
pnls[0]    # 120  (first)
pnls[-1]   # -50  (last)
pnls[1]    # -80  (second)
```

**Slicing** `list[start:stop]` returns a new list from index `start` up to but **not including** `stop`.

```python
pnls[1:3]  # [-80, 250]   — items at indices 1 and 2
pnls[:2]   # [120, -80]   — start defaults to 0
pnls[2:]   # [250, -50]   — stop defaults to end
```

**`.append(value)`** adds an item to the end of an existing list, in place.

````python
trades = [320, -150, 80, -200, 410, -90, 50]
print("first:   ", trades[0])
print("last:    ", trades[-1])
print("length:  ", len(trades))
print("slice 1-4:", trades[1:4])    # predict before running
trades.append(175)
print("after append:", trades)
````

## Part D — `for` loops and `if`

**For loop:** "do this block once for every item in the sequence":

```python
for pnl in pnls:
    print(pnl)
```

Indentation is what tells Python which lines are inside the loop. Lose the indent and you've left the loop.

**`if` / `elif` / `else`:** branch on a bool.

```python
if pnl > 0:
    label = "win"
elif pnl < 0:
    label = "loss"
else:
    label = "breakeven"
```

**Combined:** the workhorse — for every item, branch based on a property of that item.

```python
for pnl in pnls:
    if pnl > 0:
        # do something with the winners
        ...
```

````python
# Predict what gets printed BEFORE running.
pnls = [120, -80, 0, 250, -50]
for pnl in pnls:
    if pnl > 0:
        print(pnl, "win")
    elif pnl < 0:
        print(pnl, "loss")
    else:
        print(pnl, "breakeven")
````

## Part E — The accumulator pattern (the big one)

The shape:

```
[set up trackers]
for item in sequence:
    [maybe an if filter]
    [update trackers]
[use trackers after the loop]
```

Trackers carry state across iterations. The four flavors you've seen:

1. **Sum + count** → averages, ratios (e.g. avg win, profit factor).
2. **Running extreme** → max drawdown, biggest loser, MFE/MAE.
3. **List as tracker (`.append` inside loop)** → equity curve.
4. **Streak with reset** → max consecutive wins/losses, drawdown duration.
