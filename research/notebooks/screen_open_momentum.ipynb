# Open Momentum / Continuation Screen (NQ, training slice)

Pre-registration: `research/hypotheses/open_momentum.md` (locked in commit `6bc2fc2`).

**Rules of this notebook:**
- Training slice only (2018-01-02 → 2022-12-30). Test slice sealed.
- No parameter changes. Bucket count (5), opening window (09:30–10:00), primary horizon (`fwd_close`), diagnostics (`fwd_2h`, `fwd_4h`), burn-in (252), bootstrap iterations (10k), significance level (95%), and pass criteria are all fixed by the pre-registration.

```python
import sys
from pathlib import Path
from datetime import time as dt_time

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Make engine/ importable from notebook location
repo_root = Path.cwd().resolve().parents[1]
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from engine.data_loader import load_csv, split_train_test

# Locked constants from pre-registration
N_BUCKETS = 5
BURN_IN_DAYS = 252
N_BOOTSTRAP = 10_000
RNG_SEED = 20260430  # session date; reproducible bootstrap
HALF_CUTOFF = pd.Timestamp("2020-07-01")  # stability-check split point

pd.set_option("display.float_format", lambda x: f"{x:+.5%}" if abs(x) < 1 else f"{x:.4f}")
```

## 1. Load and split
Use `engine.data_loader.load_csv` + `split_train_test`. Drop the test DataFrame from memory immediately so there's no accidental exposure.

```python
DATA_PATH = repo_root / "data" / "nq_15m_data.csv"
df_full = load_csv(str(DATA_PATH))
train, _test = split_train_test(df_full)
del df_full, _test  # test slice must not leak into this notebook

print(f"Training bars: {len(train):,}")
print(f"Range: {train.index.min()}  →  {train.index.max()}")
print(f"Timezone: {train.index.tz}")
```

## 2. Extract per-day prices at the locked timestamps
From the pre-registration:
- **`p_0930`** = open of the 09:30 ET bar (RTH open).
- **`p_1000`** = open of the 10:00 ET bar (= close of the 09:45 bar; defines the end of the opening window AND the start of the continuation window).
- **`p_1200`** = open of the 12:00 ET bar (end of `fwd_2h` diagnostic horizon).
- **`p_1400`** = open of the 14:00 ET bar (end of `fwd_4h` diagnostic horizon).
- **`p_close`** = close of the 15:45 ET bar (16:00 ET print, end of primary horizon).

Days where any required bar is missing are flagged and propagate as NaN; rows with NaN on a given horizon are excluded from that horizon's stats.

```python
def price_at(train_df, target_time, col):
    m = train_df.index.time == target_time
    return pd.DataFrame({
        "date": pd.to_datetime(train_df.index.date[m]),
        "price": train_df.loc[m, col].values,
    })

p0930  = price_at(train, dt_time(9, 30),  "open").rename(columns={"price": "p_0930"})
p1000  = price_at(train, dt_time(10, 0),  "open").rename(columns={"price": "p_1000"})
p1200  = price_at(train, dt_time(12, 0),  "open").rename(columns={"price": "p_1200"})
p1400  = price_at(train, dt_time(14, 0),  "open").rename(columns={"price": "p_1400"})
p_close = price_at(train, dt_time(15, 45), "close").rename(columns={"price": "p_close"})

daily = p0930.merge(p1000, on="date", how="outer") \
             .merge(p1200, on="date", how="outer") \
             .merge(p1400, on="date", how="outer") \
             .merge(p_close, on="date", how="outer") \
             .sort_values("date").reset_index(drop=True)

n_total = len(daily)
print(f"Calendar rows (any required bar present): {n_total}")
print(f"Missing 09:30 bar: {int(daily['p_0930'].isna().sum())}")
print(f"Missing 10:00 bar: {int(daily['p_1000'].isna().sum())}")
print(f"Missing 12:00 bar: {int(daily['p_1200'].isna().sum())}")
print(f"Missing 14:00 bar: {int(daily['p_1400'].isna().sum())}")
print(f"Missing 15:45 bar: {int(daily['p_close'].isna().sum())}")
daily.head()
```

## 3. Compute `open_return` (the conditioning feature)
$$\text{open\_return}_t = \frac{p_{1000,t} - p_{0930,t}}{p_{0930,t}}$$

Signed (not absolute). One value per trading day. Days missing either bar are NaN and dropped downstream.

```python
daily["open_return"] = (daily["p_1000"] - daily["p_0930"]) / daily["p_0930"]

n_valid = int(daily["open_return"].notna().sum())
print(f"Days with valid open_return: {n_valid}")
print()
print(daily["open_return"].describe())
```

## 4. Forward returns from the 10:00 ET reference price
- `fwd_close` (PRIMARY): 10:00 → 16:00 RTH close
- `fwd_2h`   (diagnostic): 10:00 → 12:00 ET
- `fwd_4h`   (diagnostic): 10:00 → 14:00 ET

```python
daily["fwd_close"] = (daily["p_close"] - daily["p_1000"]) / daily["p_1000"]
daily["fwd_2h"]    = (daily["p_1200"]  - daily["p_1000"]) / daily["p_1000"]
daily["fwd_4h"]    = (daily["p_1400"]  - daily["p_1000"]) / daily["p_1000"]

HORIZONS = ["fwd_close", "fwd_2h", "fwd_4h"]
PRIMARY = "fwd_close"

print("Missing count per horizon:")
print(daily[HORIZONS].isna().sum())
print()
print("Unconditional stats per horizon:")
daily[HORIZONS].describe().loc[["count", "mean", "std", "min", "max"]]
```

## 5. Expanding-percentile bucketing with 252-day burn-in
For each day $i$ (with $i \ge \text{BURN\_IN\_DAYS}$), compute quintile edges from the `open_return` values on days $[0, i)$ — strictly prior. Assign day $i$ to the bucket implied by those edges. Day $i$'s bucket therefore depends only on past information.

```python
valid = daily.dropna(subset=["open_return"]).reset_index(drop=True).copy()
feat = valid["open_return"].values

buckets = np.full(len(valid), np.nan)
for i in range(BURN_IN_DAYS, len(valid)):
    history = feat[:i]
    edges = np.quantile(history, np.linspace(0, 1, N_BUCKETS + 1))
    buckets[i] = np.searchsorted(edges[1:-1], feat[i], side="right")

valid["bucket"] = buckets
scored = valid.dropna(subset=["bucket"]).copy()
scored["bucket"] = scored["bucket"].astype(int)

print(f"Burn-in days excluded: {BURN_IN_DAYS}")
print(f"Days with assigned bucket: {len(scored)}")
print()
print("Bucket counts:")
print(scored["bucket"].value_counts().sort_index().rename(index=lambda b: f"Q{b+1}"))
```

## 6. De-mean forward returns against unconditional mean
Per-horizon unconditional mean is computed on post-burn-in training days (consistent evaluation window). De-meaning strips unconditional drift so bucket means reflect *conditional* effect only.

```python
DM = {}
for h in HORIZONS:
    mu = scored[h].mean()
    DM[h] = f"{h}_dm"
    scored[DM[h]] = scored[h] - mu
    print(f"{h}: unconditional mean = {mu:+.5%}")

DM_HORIZONS = [DM[h] for h in HORIZONS]
PRIMARY_DM = DM[PRIMARY]
```

## 7. Main bucket table (full training slice, post burn-in)
5 buckets × 3 horizons. Means are de-meaned forward returns; `n` is per-bucket day count.

```python
def bucket_means(df, horizons_dm):
    rows = []
    for b in range(N_BUCKETS):
        sub = df[df["bucket"] == b]
        row = {"bucket": f"Q{b+1}", "n": len(sub)}
        for h in horizons_dm:
            row[h] = sub[h].mean()
        rows.append(row)
    return pd.DataFrame(rows)

main_table = bucket_means(scored, DM_HORIZONS)
main_table
```

```python
fig, ax = plt.subplots(figsize=(10, 6))
x = np.arange(N_BUCKETS)
for h in DM_HORIZONS:
    ax.plot(x, main_table[h].values, marker="o", label=h.replace("_dm", ""))
ax.axhline(0, color="k", linewidth=0.5, linestyle="--")
ax.set_xticks(x)
ax.set_xticklabels([f"Q{i+1}" for i in range(N_BUCKETS)])
ax.set_ylabel("De-meaned forward return")
ax.set_xlabel("open_return bucket (Q1 = biggest down-open, Q5 = biggest up-open)")
ax.set_title("NQ training slice: forward return by 30-min open_return bucket")
ax.legend()
plt.tight_layout()
plt.show()
```

## 8. Bootstrap 95% confidence intervals (day-level resampling)
For each bucket × horizon cell: 10,000 iterations, each resampling the day-level values within that bucket with replacement. The CI is the 2.5/97.5 percentiles of the resampled mean distribution. Seed is fixed for reproducibility.

```python
def bootstrap_mean_ci(values, n_iter, rng):
    v = values[~np.isnan(values)]
    if len(v) == 0:
        return (np.nan, np.nan)
    idx = rng.integers(0, len(v), size=(n_iter, len(v)))
    boot_means = v[idx].mean(axis=1)
    return tuple(np.quantile(boot_means, [0.025, 0.975]))

def bucket_ci_table(df, horizons_dm):
    rng = np.random.default_rng(RNG_SEED)
    rows = []
    for b in range(N_BUCKETS):
        sub = df[df["bucket"] == b]
        row = {"bucket": f"Q{b+1}", "n": len(sub)}
        for h in horizons_dm:
            vals = sub[h].values
            lo, hi = bootstrap_mean_ci(vals, N_BOOTSTRAP, rng)
            mean = np.nanmean(vals) if len(vals) else np.nan
            excludes_zero = (lo > 0) or (hi < 0)
            row[f"{h}_mean"] = mean
            row[f"{h}_lo"] = lo
            row[f"{h}_hi"] = hi
            row[f"{h}_sig"] = "*" if excludes_zero else ""
        rows.append(row)
    return pd.DataFrame(rows)

ci_table = bucket_ci_table(scored, DM_HORIZONS)
ci_table
```

## 9. Stability check — Half A vs Half B
Date-based split at 2020-07-01 (locked in pre-registration):
- Half A: 2018-01 → 2020-06
- Half B: 2020-07 → 2022-12

**Important**: bucket assignments for each day were computed once on the full training sequence (expanding percentiles). For the stability check we partition the already-bucketed days; we do not re-bucket within each half. This preserves the pre-registered definition.

```python
half_a = scored[scored["date"] < HALF_CUTOFF].copy()
half_b = scored[scored["date"] >= HALF_CUTOFF].copy()

print(f"Half A: {len(half_a)} days, {half_a['date'].min().date()} → {half_a['date'].max().date()}")
print(f"Half B: {len(half_b)} days, {half_b['date'].min().date()} → {half_b['date'].max().date()}")

a_table = bucket_means(half_a, DM_HORIZONS)
b_table = bucket_means(half_b, DM_HORIZONS)

print("\nHalf A bucket means:")
display(a_table)
print("\nHalf B bucket means:")
display(b_table)
```

## 10. Mechanical pass/fail evaluation
From the pre-registration, all four criteria must hold on the **`fwd_close` horizon** (primary):

1. Strictly **increasing** Q1 → Q5 means (matches momentum / continuation prediction).
2. Q1 and Q5 bootstrap 95% CIs both exclude zero.
3. Q1 mean < 0 and Q5 mean > 0 (signs match prediction).
4. Strictly increasing pattern holds in both Half A and Half B independently.

```python
def strictly_increasing(arr):
    return bool(np.all(np.diff(arr) > 0))

# Criterion 1: monotonic increasing on full slice
full_means = main_table[PRIMARY_DM].values
crit1 = strictly_increasing(full_means)

# Criterion 2: Q1 and Q5 CIs both exclude zero
q1 = ci_table.iloc[0]
q5 = ci_table.iloc[-1]
q1_excludes = (q1[f"{PRIMARY_DM}_lo"] > 0) or (q1[f"{PRIMARY_DM}_hi"] < 0)
q5_excludes = (q5[f"{PRIMARY_DM}_lo"] > 0) or (q5[f"{PRIMARY_DM}_hi"] < 0)
crit2 = bool(q1_excludes and q5_excludes)

# Criterion 3: Q1 negative, Q5 positive (continuation)
q1_mean = q1[f"{PRIMARY_DM}_mean"]
q5_mean = q5[f"{PRIMARY_DM}_mean"]
crit3 = bool((q1_mean < 0) and (q5_mean > 0))

# Criterion 4: monotonic increasing in both halves
crit4 = strictly_increasing(a_table[PRIMARY_DM].values) and strictly_increasing(b_table[PRIMARY_DM].values)

print("Pre-registered pass criteria (evaluated on fwd_close_dm, the primary horizon):\n")
print(f"  1. Strictly increasing Q1→Q5 on full slice:        {crit1}")
print(f"       means: {', '.join(f'{v:+.5%}' for v in full_means)}")
print(f"  2. Q1 and Q5 bootstrap 95% CIs exclude zero:       {crit2}")
print(f"       Q1 CI: [{q1[f'{PRIMARY_DM}_lo']:+.5%}, {q1[f'{PRIMARY_DM}_hi']:+.5%}]  excludes 0: {q1_excludes}")
print(f"       Q5 CI: [{q5[f'{PRIMARY_DM}_lo']:+.5%}, {q5[f'{PRIMARY_DM}_hi']:+.5%}]  excludes 0: {q5_excludes}")
print(f"  3. Q1 negative and Q5 positive:                    {crit3}")
print(f"       Q1 mean: {q1_mean:+.5%},  Q5 mean: {q5_mean:+.5%}")
print(f"  4. Strictly increasing in both halves:             {crit4}")
print(f"       Half A means: {', '.join(f'{v:+.5%}' for v in a_table[PRIMARY_DM].values)}")
print(f"       Half B means: {', '.join(f'{v:+.5%}' for v in b_table[PRIMARY_DM].values)}")

overall = crit1 and crit2 and crit3 and crit4
print("\n" + ("OVERALL: PASS" if overall else "OVERALL: FAIL"))
if not overall:
    binding = []
    if not crit1: binding.append("1 (monotonicity on full slice)")
    if not crit2: binding.append("2 (extreme-bucket significance)")
    if not crit3: binding.append("3 (predicted signs)")
    if not crit4: binding.append("4 (stability across halves)")
    print(f"   Binding criterion failure(s): {', '.join(binding)}")
```
